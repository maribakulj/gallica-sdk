# Search and metadata with gallica-sdk

This notebook exercises a small public Gallica workflow: one SRU search and one OAIRecord metadata lookup.

```python
from gallica import Gallica

with Gallica() as gallica:
    results = gallica.search('gallica all \"Verdun\"', maximum_records=3)
    assert results.total > 0
    assert len(results.records) > 0

    metadata = gallica.document('bpt6k5738219s').metadata()
    assert metadata.raw_xml

print('search total:', results.total)
print('first title:', results.records[0].title)
print('metadata title:', metadata.record.title)
```
