# Build a resumable corpus

This notebook validates the minimal corpus workflow and proves that a second run skips already complete artifacts.

```python
from pathlib import Path
from shutil import rmtree

from gallica import Gallica

output = Path('notebook-output/corpus')
if output.exists():
    rmtree(output)

with Gallica() as gallica:
    first = gallica.corpus(['bpt6k5738219s']).fetch(
        output,
        metadata=True,
        text=False,
        resume=True,
    )
    assert len(first.successes) == 1
    assert (output / 'bpt6k5738219s' / 'metadata.json').exists()

    second = gallica.corpus(['bpt6k5738219s']).fetch(
        output,
        metadata=True,
        text=False,
        resume=True,
    )
    assert len(second.skipped) == 1

print('first run successes:', len(first.successes))
print('second run skipped:', len(second.skipped))
```
